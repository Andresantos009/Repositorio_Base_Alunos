```python
compra = int(input('Quantos produtos deseja comprar:'))

for i in range(compra):
    produto = input('Nome do produto:')
    preco = int(input('Valor do produto:'))

    if preco >= 50:
        print(f'O produto: {produto}, com o valor de: {preco}, está caro')
    
    elif preco >= 41 and preco <= 45:
        print(f'O produto: {produto}, com o valor de: {preco}, está Ok')


    elif preco <= 40:
        print(f'O produto: {produto}, com o valor de: {preco}, está baratinho até o motoboy compra')
```
