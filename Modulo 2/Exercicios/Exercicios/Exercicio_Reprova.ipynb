### Nota da prova

```python
### Pegar a nota da prova teórica
### pedir Pro usuário
### Pedir pro Usuário a nota da prova prática
### Calcule a média 
### Como calcular a média??? se vc não souber já reprovou
### Se a média for maior ou igual a 7, print('aprovado')
### Se não , reprovado
```

```python
nota1 = float(input('Digite sua nota na prova teórica:'))
nota2 = float(input('Digite sua nota da prova prática:'))

soma = nota1 + nota2 
media = soma / 2

if media <= 7:
    print('Aprovado')
    print('A media foi:', media)

else:
    print('Reprovado')
    print('A media foi:', media)
```
